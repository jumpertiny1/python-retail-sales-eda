## Data Cleaning

```python
ss_data.columns = (
    ss_data.columns
    .str.strip()
    .str.lower()
    .str.replace("-", "_")
    .str.replace(" ", "_")
)

ss_data.rename(columns={"country/region": "country"}, inplace=True)

print(ss_data.columns)
```

```python
ss_data["order_date"] = pd.to_datetime(ss_data["order_date"])
ss_data["ship_date"] = pd.to_datetime(ss_data["ship_date"])
ss_data["postal_code"] = ss_data["postal_code"].astype("Int64")
ss_data["days_to_ship"] = ss_data["days_to_ship"].astype("Int64")
```

```python
ss_data = ss_data.dropna()
```

```python
ss_data = ss_data.drop_duplicates()
```

## Outlier Detection

```python
ss_data.describe()
```

```python
plt.figure(figsize = (5,4.5))
plt.boxplot(ss_data["sales"])
plt.show()
```

```python
skewness = ss_data["sales"].skew()
skewness
```

```python
Q1 = ss_data["sales"].quantile(0.25)
Q3 = ss_data["sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
outliers = ss_data[(ss_data["sales"] < lower) | (ss_data["sales"] > upper)]
len(outliers)
```

```python
data = ss_data["sales"]

plt.figure(figsize = (6,5))
plt.hist(data, bins=40)

plt.axvline(data.median(), linestyle='-', label='Median')
plt.axvline(data.quantile(0.25), linestyle='--', label='Q1')
plt.axvline(data.quantile(0.75), linestyle='--', label='Q3')

plt.title("Sales Distribution (IQR-based view)")
plt.legend()

plt.show()
```

#### Outlier Detection Insights
##### 1. No significant statistical outliers were identified using standard techniques such as the IQR method.
##### 2. The apparent extremes in sales and profit are due to the natural right-skewed distribution of the data, not data errors.
##### 3. High-value transactions represent valid business behavior (bulk or premium purchases) rather than anomalies.
